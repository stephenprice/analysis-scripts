```python
# get lat, lon values for mpas-o cell data
fileIn = '../mpaso-IcoswISC30E3r5-restart.nc'
ds = xarray.open_dataset(fileIn)

rad2deg = 180.0 / np.pi 

lat_mesh = ds.latCell.values * rad2deg
lon_mesh = ds.lonCell.values * rad2deg

lon_mesh_temp = np.zeros(np.shape(lon_mesh))

# convert long values to same as used in route definitions
for i in np.arange(0,np.size(lon_mesh)):
        if( lon_mesh[i] > 180 ):
            lon_mesh_temp[i] = ( lon_mesh[i] - 360 )
        else:
            lon_mesh_temp[i] = lon_mesh[i]

del lon_mesh
lon_mesh = lon_mesh_temp
del lon_mesh_temp
```

```python
#get track data (use old NSR data for development / testing)
fileIn = '../arctic_route_sections.nc'
ds = xarray.open_dataset(fileIn)

lat_A = ds.lat_A.values
lon_A = ds.lon_A.values
dis_A= ds.dis_A.values

lat_B = ds.lat_B.values
lon_B = ds.lon_B.values
dis_B= ds.dis_B.values

lat_C = ds.lat_C.values
lon_C = ds.lon_C.values
dis_C= ds.dis_C.values

lat_D = ds.lat_D.values
lon_D = ds.lon_D.values
dis_D= ds.dis_D.values

lat_E = ds.lat_E.values
lon_E = ds.lon_E.values
dis_E= ds.dis_E.values

lat_F = ds.lat_F.values
lon_F = ds.lon_F.values
dis_F= ds.dis_F.values

lat_G = ds.lat_G.values
lon_G = ds.lon_G.values
dis_G= ds.dis_G.values

lat_H = ds.lat_H.values
lon_H = ds.lon_H.values
dis_H= ds.dis_H.values

lat_I = ds.lat_I.values
lon_I = ds.lon_I.values
dis_I= ds.dis_I.values

lat_J = ds.lat_J.values
lon_J = ds.lon_J.values
dis_J= ds.dis_J.values

trans_lat=np.hstack((lat_E,lat_J,lat_C,lat_D))  
trans_lon=np.hstack((lon_E,lon_J,lon_C,lon_D))
trans_dist=np.hstack((dis_E,dis_J,dis_C,dis_D))

# trans_lat=np.hstack((lat_A,lat_B,lat_I,lat_H))  
# trans_lon=np.hstack((lon_A,lon_B,lon_I,lon_H))
# trans_dist=np.hstack((dis_A,dis_B,dis_I,dis_H))
```

```python
# overlay track and mpas-o mesh lat,lon coords
import math
from scipy.spatial import KDTree

# generates a data structure that is queryed to find nearest neighbors given set of lat, lon points
tree = KDTree( list( zip( lon_mesh.flatten(), lat_mesh.flatten() ) ) )

# using 'tree' data structure above, and passed array of lat, lon points from transect of interest,
# generates indices(ind) to nearest (k=)4 mpaso mesh cells and distance(d) between target cell and
# mpaso mesh cells (what units are distances in, degrees?)
kneighbor = 4
d, inds = tree.query( list( zip( trans_lon, trans_lat ) ), k = kneighbor )

# From LAC code: 601 refers to record time dim length (prev. study), which was 50 yrs (2050-2000) x 12 months
# Not needed here (?) ... but leaving for now for ref.

# number = inds[:,0]/601
# #do some magic here to get indices
# ind1=np.zeros((len(number)))
# ind2=np.zeros((len(number)))

# for i in np.arange(0,len(number)):
#     num=number[i]
#     def split_float(num):
#         integer_part = math.trunc(num)
#         decimal_part = num - integer_part
#         return integer_part, decimal_part

#     integer, decimal = split_float(num)
#     ind1[i]=np.int16(integer)
#     ind2[i]=np.int16(601*decimal)
```

```python
from polaris_rio import polar_rio
import numpy as np

for year in years:
    
    for month in months:
        
        fileIn = prefix+year+'-'+month+'-01.nc'
        
        # read in file and relevant vars
        print('reading IN file: ', fileIn)
        dataIn = xarray.open_dataset(fileIn)

        # ice area
        iceAreaMedian = dataIn.timeDaily_avg_iceAreaCell_ensembleMedian.values
        iceArea5th = dataIn.timeDaily_avg_iceAreaCell_ensemble5th.values
        iceArea95th = dataIn.timeDaily_avg_iceAreaCell_ensemble95th.values

        # ice thickness
        iceThickMedian = dataIn.timeDaily_avg_iceVolumeCell_ensembleMedian.values
        iceThick5th = dataIn.timeDaily_avg_iceVolumeCell_ensemble5th.values
        iceThick95th = dataIn.timeDaily_avg_iceVolumeCell_ensemble95th.values
                
        #ndays = np.size( iceThickMedian, axis=0 )     # production: loop over days
        ndays = np.arange(15,16)                     # debug: use single day
        ncells = np.size( iceThickMedian, axis=1 )
        
        for day in ndays:
                        
            print('calculating RIO information for day: ', day)
                
            # Use indices from tree query above to extract and calculate RIO value at
            # points along route of interest. 
            RIO_PC3 = np.ones(len(trans_dist))*30
            RIO_PC3_5th = np.ones(len(trans_dist))*30
            RIO_PC3_95th = np.ones(len(trans_dist))*30
            RIO_PC6 = np.ones(len(trans_dist))*30
            RIO_PC6_5th = np.ones(len(trans_dist))*30
            RIO_PC6_95th = np.ones(len(trans_dist))*30
            RIO_NIS = np.ones(len(trans_dist))*30
            RIO_NIS_5th = np.ones(len(trans_dist))*30
            RIO_NIS_95th = np.ones(len(trans_dist))*30
        
            # temp thick and conc. vectors for debugging / plotting
            thickAlong = np.zeros(len(trans_dist))
            areaAlong = np.zeros(len(trans_dist))
            thickAlong5th = np.zeros(len(trans_dist))
            areaAlong5th = np.zeros(len(trans_dist))
            thickAlong95th = np.zeros(len(trans_dist))
            areaAlong95th = np.zeros(len(trans_dist))

            for transPoint in np.arange(0,np.size(trans_dist)):
        
                # DEBUG: pull index into mpas data vector associated w/ closest point to transect value
                # NOT NEEDED IF USING MEADIAN OVER NNs
#                 indMinDist = np.argmin(d[transPoint,:])
#                 mpasInd = inds[transPoint,indMinDist]
                
                # calc. average median thickness (over neighboring cells) at transect point 
                thickInMedian = np.mean( iceThickMedian[day,inds[transPoint,0:kneighbor]] )
                thickAlong[transPoint]=thickInMedian

                # calc. average median conc. (over neighboring cells) at transect point 
                areaInMedian = np.mean( iceAreaMedian[day,inds[transPoint,0:kneighbor]] )
                areaAlong[transPoint]=areaInMedian

                # calc. average 5th percentile thickness (over neighboring cells) at transect point 
                thickIn5th = np.mean( iceThick5th[day,inds[transPoint,0:kneighbor]] )
                thickAlong5th[transPoint]=thickIn5th

                # calc. average 5th percentile conc. (over neighboring cells) at transect point 
                areaIn5th = np.mean( iceArea5th[day,inds[transPoint,0:kneighbor]] )
                areaAlong5th[transPoint]=areaIn5th
                
                # calc. average 95th percentile thickness (over neighboring cells) at transect point 
                thickIn95th = np.mean( iceThick95th[day,inds[transPoint,0:kneighbor]] )
                thickAlong95th[transPoint]=thickIn95th

                # calc. average 95th percentile conc. (over neighboring cells) at transect point 
                areaIn95th = np.mean( iceArea95th[day,inds[transPoint,0:kneighbor]] )
                areaAlong95th[transPoint]=areaIn95th
                

                # use the product of these to calculate RIO values
                                
                pclass='PC3'       
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                                 ice_thickness_m=thickInMedian, 
                                                 concentration_tenths=areaInMedian*10 
                                                )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_PC3[transPoint] = rio
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                             ice_thickness_m=thickIn5th, 
                                             concentration_tenths=areaIn5th*10 
                                            )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_PC3_5th[transPoint] = rio
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                             ice_thickness_m=thickIn95th, 
                                             concentration_tenths=areaIn95th*10 
                                            )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_PC3_95th[transPoint] = rio

                
                pclass='PC6'                
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                                 ice_thickness_m=thickInMedian, 
                                                 concentration_tenths=areaInMedian*10 
                                                )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_PC6[transPoint] = rio
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                             ice_thickness_m=thickIn5th, 
                                             concentration_tenths=areaIn5th*10 
                                            )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_PC6_5th[transPoint] = rio
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                             ice_thickness_m=thickIn95th, 
                                             concentration_tenths=areaIn95th*10 
                                            )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_PC6_95th[transPoint] = rio
                
                
                pclass='Not Ice Strengthened'                
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                                 ice_thickness_m=thickInMedian, 
                                                 concentration_tenths=areaInMedian*10 
                                                )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_NIS[transPoint] = rio
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                             ice_thickness_m=thickIn5th, 
                                             concentration_tenths=areaIn5th*10 
                                            )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_NIS_5th[transPoint] = rio
                rio, ice_type, level = polar_rio( vessel_ice_class=pclass, 
                                             ice_thickness_m=thickIn95th, 
                                             concentration_tenths=areaIn95th*10 
                                            )                                             
                if( (thickInMedian > 0.0) & (areaInMedian > 0.0) ):
                    RIO_NIS_95th[transPoint] = rio
```
