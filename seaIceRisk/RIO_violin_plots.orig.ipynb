```python
from netCDF4 import Dataset
import numpy as np
import matplotlib.pyplot as plt
import xarray
import glob, os

#get track data
path='/users/lconlon/'
os.chdir(path)
file='arctic_route_sections.nc'

ds = xarray.open_dataset(file)
lat_A = ds.lat_A.values
lon_A = ds.lon_A.values
dis_A= ds.dis_A.values

lat_B = ds.lat_B.values
lon_B = ds.lon_B.values
dis_B= ds.dis_B.values

lat_C = ds.lat_C.values
lon_C = ds.lon_C.values
dis_C= ds.dis_C.values

lat_D = ds.lat_D.values
lon_D = ds.lon_D.values
dis_D= ds.dis_D.values

lat_E = ds.lat_E.values
lon_E = ds.lon_E.values
dis_E= ds.dis_E.values

lat_F = ds.lat_F.values
lon_F = ds.lon_F.values
dis_F= ds.dis_F.values

lat_G = ds.lat_G.values
lon_G = ds.lon_G.values
dis_G= ds.dis_G.values

lat_H = ds.lat_H.values
lon_H = ds.lon_H.values
dis_H= ds.dis_H.values

lat_I = ds.lat_I.values
lon_I = ds.lon_I.values
dis_I= ds.dis_I.values

lat_J = ds.lat_J.values
lon_J = ds.lon_J.values
dis_J= ds.dis_J.values
```

```python
import math
from scipy.spatial import cKDTree

#use this one for the seasonal violin plot with no colors

#******************************
# Change these things if you want:
trans='IBA'          #Select transect
trans_lat=np.hstack((lat_I,lat_B,lat_A))   #Also have to select transect here
trans_lon=np.hstack((lon_I,lon_B,lon_A))
trans_dist=np.hstack((dis_I,dis_B,dis_A))
use_mean=False     #true=plot 5 year means; false= plot every year
use_seasons=True
season=1 #used only when use_seasons is true. Spring=0, Summer=1, Fall=2, Winter=3
PC=5               #select polar class here
#******************************


tree = cKDTree(list(zip(lon.flatten(), lat.flatten())))
d, inds = tree.query(list(zip(trans_lon, trans_lat)), k = 4)


number = inds[:,0]/601
#do some magic here to get indices
ind1=np.zeros((len(number)))
ind2=np.zeros((len(number)))

for i in np.arange(0,len(number)):
    num=number[i]
    def split_float(num):
        integer_part = math.trunc(num)
        decimal_part = num - integer_part
        return integer_part, decimal_part


    integer, decimal = split_float(num)
    ind1[i]=np.int16(integer)
    ind2[i]=np.int16(601*decimal)

    
plot_data=(ROI_ens1+ROI_ens2+ROI_ens3+ROI_ens4+ROI_ens5)/5 #ensemble mean

#Get ROIs on transect indices
plot_data=plot_data[:,:,np.int16(ind1),np.int16(ind2)]
# Fill in NaN's...
mask = np.isnan(plot_data)
plot_data[mask] = np.interp(np.flatnonzero(mask), np.flatnonzero(~mask), plot_data[~mask])

speeds=np.zeros((50,12,len(trans_dist)))

#now set speeds. For all vessels:
#speeds[plot_data>0]=19
speeds[(plot_data <-10)]=np.nan

#limited operations
if PC==1:
    speeds[plot_data>0]=19
    speeds[(plot_data <0) & (plot_data > -10)]=11 # PC1
if PC==2:
    speeds[plot_data>0]=16.5
    speeds[(plot_data <0) & (plot_data > -10)]=8 # 
if PC==3:
    speeds[plot_data>0]=16
    speeds[(plot_data <0) & (plot_data > -10)]=5 # 
if PC==4:
    speeds[plot_data>0]=15
    speeds[(plot_data <0) & (plot_data > -10)]=5 # 
if PC==5:
    speeds[plot_data>0]=14
    speeds[(plot_data <0) & (plot_data > -10)]=5 # 
if PC==6:
    speeds[plot_data>0]=14
    speeds[(plot_data <0) & (plot_data > -10)]=3 # 3 knots for PC6
if PC==7:
    speeds[plot_data>0]=14
    speeds[(plot_data <0) & (plot_data > -10)]=3 # 3 knots for PC6

#calc transit time based on speeds
time=1/speeds
time=np.sum(time,2)/24
masked_data = np.ma.masked_where(np.isinf(time), time)
###########################################################

year1=2000 #np.int16(sorted(glob.glob("*clim*"))[0][5:9])
yearend=2050 #p.int16(year1)+len(ROI_ens1[:,0,0,0])
all_years_no_mean=np.arange(year1,yearend,1)
all_years_mean=np.arange(year1,yearend,5)


all_data_no_mean = time.T 

#five year means
all_data_mean=np.zeros((12,np.int16(len(all_data_no_mean[0,:])/5)))
all_data_mean[:,0]=np.nanmean(all_data_no_mean[:,0:4],1)
all_data_mean[:,1]=np.nanmean(all_data_no_mean[:,5:9],1)
all_data_mean[:,2]=np.nanmean(all_data_no_mean[:,10:14],1)
all_data_mean[:,3]=np.nanmean(all_data_no_mean[:,15:19],1)
all_data_mean[:,4]=np.nanmean(all_data_no_mean[:,20:24],1)
all_data_mean[:,5]=np.nanmean(all_data_no_mean[:,25:29],1)
all_data_mean[:,6]=np.nanmean(all_data_no_mean[:,30:34],1)
all_data_mean[:,7]=np.nanmean(all_data_no_mean[:,35:39],1)
all_data_mean[:,8]=np.nanmean(all_data_no_mean[:,40:44],1)
all_data_mean[:,9]=np.nanmean(all_data_no_mean[:,45:],1)

if use_mean==True:
    all_data=all_data_mean
    all_years=all_years_mean
if use_mean==False:
    all_data=all_data_no_mean
    all_years=all_years_no_mean

summer=all_data[5:8,:] #JJA
fall=all_data[8:11,:] #SON
spring=all_data[2:5,:] #MAM
winter1=all_data[11,:]
winter2=all_data[0:2,:]
winter=np.vstack((winter1,winter2))
seasons=np.dstack((spring,summer, fall, winter))
if use_seasons==True:
    all_data=seasons[:,:,season]
if use_seasons==False:
    all_data=all_data
fig, axs = plt.subplots(nrows=1, ncols=1, figsize=(12, 4))
# plot violin plot 1
for i in np.arange(0,len(all_data[0,:])):
    parts=axs.violinplot(all_data[~np.isnan(all_data[:,i]),i], positions=[i+1],
                      showmeans=True,
                      showmedians=True,showextrema=True)
    ymin=np.nanmin(all_data[:,i])
    ymax=np.nanmax(all_data[:,i])
    ymean=np.nanmean(all_data[:,i])

    
    for pc in parts['bodies']:
        number=sum(~np.isnan(all_data[:,i]))
        pc.set_facecolor('deepskyblue') ###
        pc.set_alpha(.5)
        axs.hlines(ymean, [i+1-0.2], [i+1.2], color='deepskyblue',linewidth=1.1)
        axs.hlines(ymin, [i+1-0.2], [i+1.2], color='deepskyblue',linewidth=1.1)
        axs.hlines(ymax, [i+1-0.2], [i+1.2], color='deepskyblue',linewidth=1.1)
        axs.vlines(i+1, ymin, ymax, color='deepskyblue',linestyle='-',linewidth=1.1)



# set labels, etc
axs.set_title('Transect '+ trans+', PC'+str(PC))
axs.set_xticks([y + 1 for y in range(len(all_data.T))],
                  labels=all_years, rotation=90)
axs.set_xlabel('Year')
axs.set_ylabel('Transit time, Days')
axs.set_ylim([np.nanmin(time),np.nanmax(time)+0.25])
path='/users/lconlon'
os.chdir(path)
    #plt.savefig('C_violin.png')
plt.show()
```
